```python
import time
from ultralytics import YOLO
model = YOLO('yolov8s.pt')

time_start = time.time()
# Train the model
results = model.train(
    data='dataset.yaml',
    epochs=20,
    imgsz=640,
    batch=4,
    device='mps' # Keep using 'mps' for your Apple M1!
)

time_end = time.time()

print(f"total time taken in the train is : {time_end - time_start}")
```

```python
from ultralytics import YOLO

# 1. Load your newly trained model
trained_model_path = '/Users/yash/Library/artikate/Artikate_Assignment/scripts/runs/detect/train-8/weights/best.pt' # Check your actual path!
model = YOLO(trained_model_path)

# 2. Export the model to ONNX format (FP32 precision)
# This will create a file named 'best.onnx' in the same folder as best.pt
onnx_path = model.export(format='onnx', dynamic=False)
print(f"ONNX model saved to: {onnx_path}")
```

```python
import cv2
import numpy as np
from ultralytics import YOLO

# Load models
pt_model = YOLO('/Users/yash/Library/artikate/Artikate_Assignment/scripts/runs/detect/train-8/weights/best.pt')
onnx_model = YOLO('/Users/yash/Library/artikate/Artikate_Assignment/scripts/runs/detect/train-8/weights/best.onnx', task='detect')

img_path = '/Users/yash/Library/artikate/Artikate_Assignment/scripts/images/val/IMG20260906153940.jpg'

# Run inference
res_pt = pt_model(img_path)[0]
res_onnx = onnx_model(img_path)[0]

# Plot results
img_pt = res_pt.plot()
img_onnx = res_onnx.plot()

# Save to artifacts dir
cv2.imwrite('/Users/yash/Library/artikate/Artikate_Assignment/scripts/results/pytorch_result.jpg', img_pt)
cv2.imwrite('/Users/yash/Library/artikate/Artikate_Assignment/scripts/results/onnx_result.jpg', img_onnx)
print("Images saved.")
```

```python
import numpy as np

# Pick a validation image to test
test_image = '/Users/yash/Library/artikate/Artikate_Assignment/scripts/images/val/IMG20260906153940.jpg' # Use any image from your val set

# Run inference with PyTorch (.pt)
pytorch_results = model(test_image)[0]
pytorch_boxes = pytorch_results.boxes.xyxy.cpu().numpy()

# Run inference with ONNX (.onnx)
onnx_model = YOLO(onnx_path)
onnx_results = onnx_model(test_image)[0]
onnx_boxes = onnx_results.boxes.xyxy.cpu().numpy()

# Compare the output boxes mathematically
print("\n--- Match Confirmation ---")
if len(pytorch_boxes) == len(onnx_boxes):
    # Check if the coordinates are almost exactly the same (floating point math can vary slightly)
    max_diff = np.max(np.abs(pytorch_boxes - onnx_boxes))
    print(f"Maximum difference in bounding box coordinates: {max_diff}")
    if max_diff < 1e-3:
        print("✅ The ONNX output matches the PyTorch output!")
    else:
        print("❌ The outputs differ slightly.")
else:
    print("❌ The models detected a different number of objects.")
```
